```
# conda install conda-forge::psycopg2-binary
import psycopg2
```

```
def test_database_login(host, port, database, user, password):
    try:
        conn = psycopg2.connect(host=host, port=port, database=database, user=user, password=password)
        print("Connection successful")
        conn.close()

    except Exception as error:
        print("Connection failed")
        print(error)

test_database_login("localhost", "5432", "test_db", "postgres", "dani2")
```

```
def iteratively_add_to_database(host, port, database, user, password):
    try:
        conn = psycopg2.connect(host=host, port=port, database=database, user=user, password=password)
        cursor = conn.cursor()

        for i in range(10000000):
            cursor.execute("INSERT INTO test_table (entry_date, observation) VALUES (%s, %s)",
                           ("2026-03-10", i))

        conn.commit()
        print("Data added successfully")

        cursor.close()
        conn.close()

    except Exception as error:
        print("Failed to add data")
        print(error)

iteratively_add_to_database("localhost", "5432", "test_db", "postgres", "dani2")
```

```
def example_query(host, port, database, user, password):
    try:
        conn = psycopg2.connect(host=host, port=port, database=database, user=user, password=password)
        cursor = conn.cursor()

        query = "SELECT * FROM test_table WHERE entry_date = %s AND observation = %s"
        cursor.execute(query, ("2026-03-10", 22))

        rows = cursor.fetchall()
        for row in rows:
            print(row)

        conn.close()

    except Exception as error:
        print("Failed to add data")
        print(error)

example_query("localhost", "5432", "test_db", "postgres", "dani2")
```
